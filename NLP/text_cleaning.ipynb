```python
import re
```

```python
def html_removal(text):
    empty_html = []
    for i in text:
        clean = re.compile('<.*?>')
        empty_html.append(re.sub(clean, '', i))
    return empty_html
```

```python
def url_removal(text):
    empty_url = []
    for i in text:
        clean = re.compile(r'http\S+')
        empty_url.append(re.sub(clean, '', i))
    return empty_url
```

```python
def special_char_removal(text):
    empty_special = []
    for i in text:
        empty_special.append(re.sub(r"[^a-zA-Z]", "", i))
    return empty_special
```

```python
def removal_punctuations(text):
    empty_punct = []
    for i in text:
        empty_punct.append(re.sub(r'[^\w\s]', '', i))
    return empty_punct
```

```python
slang_dict = {
    "ain't": "am not",
    "aren't": "are not",
    "can't": "cannot",
    "could've": "could have",
    "couldn't": "could not",
    "didn't": "did not",
    "doesn't": "does not",}
```

```python
def expand_slang(text,slang_dict):
    expanded_text = []
    for i in text:
        for word in slang_dict:
            if word in i:
                i = i.replace(word, slang_dict[word])
        expanded_text.append(i)
    return expanded_text
```
